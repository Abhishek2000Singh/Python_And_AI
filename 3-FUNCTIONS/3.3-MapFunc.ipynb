```python
numbers=[1,2,3,4,5,6]

list(map(lambda x:x**2,numbers))
```

```python
def square(nums):
    return nums**2

square(4)
```

```python
list(map(square,numbers))
```

```python
num1 = [1,2,3,4]
num2=[5,6,7,8]

added_num = list(map(lambda x,y:x+y,num1,num2))
print(added_num)
```

```python
## Map to convert the list of string to integer

str_num = ['1','2','3']
int_num = list(map(int,str_num))
print(int_num)
```

```python
def get_name(person):
    return person['name']

people=[
    {'name':'Abhishek','age':26},
    {'name':'Harshit','age':23},
]

list(map(get_name,people))
```
