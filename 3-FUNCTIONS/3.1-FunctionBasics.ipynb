```python
def even_or_Odd(num):
    if(num%2==0):
        print(f"{num}" + " is even")
    else:
        print(f"{num}" + " is Odd")
```

```python
### Variable length Arguments or Positional and Keywords Arguement
def print_num(*args):
    for num in args:
        print(num)
    
```

```python
print_num(1,2,3,4,5,"GenZ")
```

```python
##KeyWords Arguement : all params/args will be in form of keyvalue forms

def print_details(**kwargs):
    for key,value in kwargs.items():
        print(f"{key}:{value}")
```

```python
print_details(name="Abhishek",age=26,country="India")
```

```python
##KeyWords Arguement : all params/args will be in form of keyvalue forms

def print_details(*args,**kwargs):
    for  val in args:
        print(f" Positional Arg: {val}")

    for key,value in kwargs.items():
        print(f"{key}:{value}")

    for key in kwargs.keys():
        print(f"Keys Only:{key}")
```

```python
print_details(1,2,3,4,5,"Harsh",name="Abhishek",age=26,country="India")
```
