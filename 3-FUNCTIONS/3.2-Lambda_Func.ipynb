# Lamba Function

```python
## Syntax
lambda arguements : expressions
```

```python
def add(a,b):
    return a+b
    
```

```python
add(2,3)
```

```python
addi = lambda a,b : a+b

type(addi)
```

```python
print(addi(3,6))
```

```python
even1 = lambda num:num%2==0
even1(12)
```
