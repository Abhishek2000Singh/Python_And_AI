```python
##Abstraction means hiding complex data or implementation details and showing only the essential features of the object. It helps to reduce programming complexity and effort. In Python, abstraction can be achieved through abstract classes and methods.

##Encapsulation means restricting access to certain attributes and methods of an object. It helps to protect the internal state of the object from unintended interference and misuse. In Python, encapsulation can be achieved through private and protected attributes and methods.
```

```python
## Encapsulation and abstraction are two important concepts in object-oriented programming that help to improve code organization, maintainability, and security.
## Public,protected,private variable i.e access modifiers
```

```python
class Person:
    def __init__(self,name,age):
        self.name = name
        self.age= age


def get_name(self):
        return self.name

person = Person("Abhishek",22)
print(get_name(person))
```

```python
class Person:
    def __init__(self,name,age,gender):
        self.__name = name  ## __ to define private variable
        self._age= age ## _ to define protected variable
        self.gender = gender

class Employee(Person):
      def __init__(self,name,age,gender):
            super().__init__(name,age,gender)

def get_name(self):
        return self.__name

def get_age(self):
        return self._age

def get_gender(self):
        return self.gender

```
