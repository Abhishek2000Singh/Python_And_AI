## Abstraction 

ABSTRACTION is the concept of hiding the complex reality while exposing only the essential parts. It is a fundamental principle in object-oriented programming (OOP) that allows developers to create simplified models of complex systems. By using abstraction, programmers can focus on high-level functionalities without getting bogged down by the intricate details of implementation.

```python
from abc import ABC, abstractmethod

## Abstract class for all the different type of vehicle
class Vehicle(ABC):

    def driver(self):
        print("The driver is driving the vehicle.")

    @abstractmethod
    def start(self):
        pass

    @abstractmethod
    def stop(self):
        pass

    @abstractmethod
    def get_fuel_type(self):
        pass


class Car(Vehicle):

    def start(self):
        print("The car is starting.")

    def stop(self):
        print("The car is stopping.")

    def get_fuel_type(self):
        return "Petrol"


def operate_veh(vehicle):
    print(vehicle.start())
    print(vehicle.get_fuel_type())
    vehicle.driver()


car = Car()
operate_veh(car)
```
