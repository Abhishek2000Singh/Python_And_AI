```python
### A class is a blueprint for creating objects. It defines a set of attributes and methods that the created objects will have. In Python, you can define a class using the `class` keyword, and you can create instances (objects) of that class.

class Car:
    pass

audi= Car()  # Creating an instance of the Car class
bmw= Car()   # Creating another instance of the Car class

print(type(audi))  # Output: <class '__main__.Car'>
print(type(bmw))   # Output: <class '__main__.Car'>
```

```python
print(audi)
```

```python
audi.windows= 4
print(audi.windows)  # Output: 4
```

```python
### Instance variables are specific to each instance of a class. They are defined within the `__init__` method and are prefixed with `self`. Each object can have different values for these instance variables.

dir(audi) # This will show the attributes of the audi object, including 'windows'
```

```python
class Dog:
    ##Constructors and Instance Variables
    def __init__(self, name, breed):
        self.name = name
        self.breed = breed

# Creating instances of the Dog class
dog1 = Dog("Buddy", "Golden Retriever")
dog2 = Dog("Max", "German Shepherd")

# Accessing instance variables
print(dog1.name)  # Output: Buddy
print(dog1.breed)  # Output: Golden Retriever
print(dog2.name)  # Output: Max
print(dog2.breed)  # Output: German Shepherd
```

```python
## Define a class with instance method
class Dog:
    def __init__(self, name, breed):
        self.name = name
        self.breed = breed

    def bark(self):
        print(f"{self.name} says Woof!")
```

```python
dog1 = Dog("Buddy", "Golden Retriever")
dog1.bark() # Output: Buddy says Woof!

dog2 = Dog("Max", "German Shepherd")
dog2.bark() # Output: Max says Woof!
print(dog2.name)
```

```python
## Modeling a Bank Account
class BankAccount:
    def __init__(self,owner, balance=0):
        self.owner = owner
        self.balance = balance

    def deposit(self, amount):
        if amount > 0:
            self.balance += amount
            print(f"Deposited {amount}. New balance is {self.balance}.")
        else:
            print("Deposit amount must be positive.")

    def withdraw(self, amount):
        if 0 < amount <= self.balance:
            self.balance -= amount
            print(f"Withdrew {amount}. New balance is {self.balance}.")
        else:
            print("Invalid withdrawal amount or insufficient funds.")

    def get_balance(self):
        return self.balance

```

```python
bank_account = BankAccount("Alice", 1000)
bank_account.deposit(500)  # Deposited 500. New balance is 1500
bank_account.withdraw(200)  # Withdrew 200. New balance is 1300
bank_account.withdraw(2000)  # Invalid withdrawal amount or insufficient funds.
bank_account.get_balance()  # Output: 1300
```
