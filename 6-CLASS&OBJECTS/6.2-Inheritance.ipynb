```python
## Inheritance
##Parent class

class Car:
    def __init__(self, windows, door, engineType):
        self.windows = windows
        self.door = door
        self.engineType = engineType

    def drive(self):
        print(f"The person will driver the car with {self.windows} windows, {self.door} doors, and an {self.engineType} engine.")
    
```

```python
car1 = Car(4, 4, "V6")
car1.drive()
```

```python
class Tesla(Car):
    def __init__(self, windows, door, engineType, autopilot):
        super().__init__(windows, door, engineType)
        self.autopilot = autopilot

    def selfdrive(self):
        print(f"The person will driver the Tesla with {self.windows} windows, {self.door} doors, an {self.engineType} engine, and autopilot is {self.autopilot}.")
```

```python
car2 = Tesla(4, 4, "Electric", "Supported")
car2.drive()
```

```python
## Multiple Inheritance -- when a class inherits from more than one parent class, it is called multiple inheritance. The child class inherits the properties and methods of all the parent classes.

## Base Class 1
class Animal:
    def __init__(self, name):
        self.name = name

    def speak(self):
        print("Subclasses must implement this method.")

## Base Class 2
class Pet:
    def __init__(self, owner):
        self.owner = owner

    def speak(self):
        print(f"{self.name} is playing with {self.owner}.")

##Derived Class
class Dog(Animal, Pet):
    def __init__(self, name, owner):
        # super().__init__(name)
        # super(Pet, self).__init__(owner) this will throw an error because super() can only be used to call the parent class's method, and in this case, we have two parent classes. Instead, we can call the __init__() method of each parent class directly.
        Animal.__init__(self, name) 
        Pet.__init__(self, owner)

    def speak(self):
        print(f"{self.name} says Woof!")
```

```python
dog = Dog("Buddy", "Alice")
dog.speak()  # Output: Buddy says Woof!
print(f"{dog.name} is playing with {dog.owner}.")  # Output: Buddy is playing with Alice.
```
