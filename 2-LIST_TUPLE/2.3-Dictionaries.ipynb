## Dictionaries

Dictionaries are unordered collection of items. They store in key-value pairs. Key must be unique and immutable, while value can be of any type

```python
## Creating Dictionaries
empty_dict = {}
print(type(empty_dict))
```

```python
empty_dict = dict()
empty_dict
```

```python
student = {"name": "Abhishek","age":32,"grade":"12th"}
print(student)
print(type(student))
```

```python
## Error

student = {"name": "Abhishek","age":32,"name":"12th"}
print(student)
print(type(student))
```

```python
## acess

student = {"name": "Abhishek","age":32,"grade":"12th"}
student['grade']
```

```python
print(student['age'])

##Acess using get method
print(student.get('grade'))
print(student.get('aged')) ## will return none
print(student.get('age'))
print(student.get('last_name',"Not Available"))## default value provided
```

```python
## Modifying dictionary element
## Dictionaries are mutable

print(student)
```

```python
student["age"]=33  ## updated
print(student)
student["address"]="India" ## added new key-value
print(student)
```

```python
## delete
del student["grade"]
print(student)
```

```python
##Methods

keys = student.keys() ## get all keys
print(keys)
value = student.values()
print(value)

items = student.items() ## list of tuple
print(items)
```

```python
## Shalow Copy
student_copy = student
print(student)
print(student_copy)
```

```python
student["name"] = "Harshit"
print(student)
print(student_copy)
```

```python
student_copy1 = student.copy() ##deep copy
print(student_copy1)
print(student)
```

```python
student["name"]= "Abhishek"
print(student)
print(student_copy1)
```

```python
## Iterating over Dictionaries
## You can loops to Iterating over Dictionaries
##Iterating over key

for keys in student.keys():
    print(keys)
```

```python
for key,value in student.items():
    print(f"{key}-{value}")
```

```python
## Nested Dictionaries
students = {
  "student1":{"name":"Abhishek","age":26},
  "student2":{"name":"Harshit","age":23},
 }
print(students)
print(students.items())
```

```python
## Acessing nested dictionaries
print(students["student2"]["name"])
print(students["student2"]["age"])
```

```python
## iterating over nested

for student_id,student_info in students.items():
    print(f"{student_id}:{student_info}")
    for key,value in student_info.items():
        print(f"{key}:{value}")


```

```python
## Dictionary Comprehension

squares={x:x**2 for x in range(5)}
print(squares)
```

```python
## Condtion Comprehension

evens = {x:x**2 for x in range(10) if x%2==0}
print(evens)
```

```python
## Practical Example

## Use a dictionary to count the frequency of elements in list
numbers = [1,2,2,3,3,3,4,4,4,4]
frequency ={}


for num in numbers:
    if num in frequency:
        frequency[num]+=1
    else:frequency[num]=1

print(frequency)
```

```python
## Merge two dictionaries

dict1= {"a":1,"b":2}
dict2= {"b":3,"c":4}

merge_dict = {**dict1,**dict2}
print(merge_dict)
```
