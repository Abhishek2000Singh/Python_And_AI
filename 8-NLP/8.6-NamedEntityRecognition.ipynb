```python
sentence= "The Effel Tower was built from 1887 to 1889 by French Engineer Gustave Eiffel, whose company specialised in building metal frameworks and structures"
```

```python
import nltk
from nltk.corpus import gutenberg
nltk.download('words')
nltk.download('maxent_ne_chunker')
nltk.download('maxent_ne_chunker_tab')
```

```python
words = nltk.word_tokenize(sentence)
```

```python
tag_element = nltk.pos_tag(words)
```

```python
nltk.ne_chunk(tag_element).draw()
```
