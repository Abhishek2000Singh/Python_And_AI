```python
## SPeech of Dr APJ ABDUL KALAM
paragraph= """
Good morning to everyone.
Today, I would like to speak about the importance of dreams, education, and hard work. A dream is not just something that we see when we are asleep. It is something that gives us the courage to work for a better future.We should always have a dream, and we should work hard to achieve it. In life, there will be many challenges, but we must not be afraid of them. With determination and confidence, we can overcome the difficulties that come in our way.
Education is one of the most powerful tools in the world. It gives us knowledge, and it helps us to understand the world around us. The young people of our country have the power to bring great changes to society.
We must believe in ourselves and in our abilities. We should use our knowledge for the development of our nation and for the benefit of others. If we work together with honesty, discipline, and dedication, we can make our country stronger and better.So, let us dream big, let us learn every day, and let us work with courage and determination. The future is in our hands, and it is our responsibility to make it bright.
Thank you.
"""
```

```python
from nltk.stem import PorterStemmer
```

```python
from nltk.corpus import stopwords
```

```python
import nltk
nltk.download('stopwords')
```

```python
stopwords.words('english')
```

```python
from nltk.stem import PorterStemmer
```

```python
stemmer = PorterStemmer()
```

```python
sentences = nltk.sent_tokenize(paragraph)
type(sentences)
```

```python
## Apply stopwords and filter and then apply stemming

for i in range(len(sentences)):
    words = nltk.word_tokenize(sentences[i])
    words =[stemmer.stem(word) for word in words if word not in set(stopwords.words('english')) ]
    sentences[i] = ' '.join(words)  ## converting all the list of  words into sentences
```

```python
sentences
```

```python
from nltk.stem import SnowballStemmer
snow_ball = SnowballStemmer('english')
```

```python
## Apply stopwords and filter and then apply snowball stemming

for i in range(len(sentences)):
    words = nltk.word_tokenize(sentences[i])
    words = [snow_ball.stem(word) for word in words if word not in set(stopwords.words('english'))]
    sentences[i] = ' '.join(words)   ## converting all list of words into sentences
```

```python
sentences
```

```python
from nltk.stem import WordNetLemmatizer
lemmetizer = WordNetLemmatizer()
```

```python
## Apply stopwords and filter and then apply snowball stemming

for i in range(len(sentences)):
    words = nltk.word_tokenize(sentences[i])
    words = [lemmetizer.lemmatize(word.lower(),pos='v') for word in words if word not in set(stopwords.words('english'))]
    sentences[i] = ' '.join(words)   ## converting all list of words into sentences
```
