```python
## Stemming is process of reducing a word to its word stem that that affixes to suffixes and prefixes or to the root if words known as lemma 

## Classification Probelm 
## Comment of Product is a positive or negative review
## Reviews --->  eating, eat, eaten here EAT is stem word or root word which will not impact much which is basically talking about EAT, each words represent vector--> eat
```

```python
words = ["eating","eats","eaten","writing","writes","programming","pragrams","history","finalized","finally"]
```

## Porter Stemmer

```python
from nltk.stem import PorterStemmer
```

```python
stemming = PorterStemmer()
```

```python
for word in words:
    print(word+"--->"+stemming.stem(word))
```

```python
stemming.stem('congratulations') ## there is major disadvantage as it changes meaning most of the times
```

```python
## RegexpStemmer Class for regular Expression
```

```python
from nltk.stem import RegexpStemmer
```

```python
reg_stemmer = RegexpStemmer('ing$|s$|e$|able$', min=4)
```

```python
reg_stemmer.stem('eating')
```

```python
reg_stemmer.stem('readable')
```

## Snowball Stemmer

```python
## better accuracy 
from nltk.stem import SnowballStemmer
```

```python
snow_ball = SnowballStemmer('english')
```

```python
for word in words:
    print(word+'--->'+snow_ball.stem(word))
```

```python
stemming.stem('fairly'),stemming.stem('sportingly')
```

```python
snow_ball.stem('fairly'),snow_ball.stem('sportingly')
```

```python
snow_ball.stem('goes'), stemming.stem('goes')
```
